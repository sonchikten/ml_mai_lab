10. Как конвертировать тип массив от float32 в int32 in place?

````python
arr = np.array([1.5, 2.7, 3.1], dtype=np.float32)

arr_int = arr.view(np.int32)
arr_int[:] = arr
print(arr_int, arr_int.dtype)
````

Метод view() создаёт новое представление массива без копирования данных. Массивы arr и arr_int используют одну и ту же область памяти, но интерпретируют её по-разному: как float32 и int32 соответственно.
При присваивании arr_int[:] = arr значения преобразуются в int32 и записываются в общую память.
В результате arr_int содержит [1, 2, 3], однако исходный массив arr также изменяется на уровне памяти. Его тип остаётся float32, поэтому при обращении к нему значения будут интерпретироваться некорректно.
Таким образом, выполняется изменение данных без выделения нового буфера памяти, но с побочным эффектом в виде повреждения исходного представления массива.

5. Правда ли, что те, кто спит больше(75 перцентиль и выше) сдает экзамен лучше, чем студенты с недосыпом(25 перцентиль и меньше)

````python
p25, p75 = df["Sleep_Hours"].quantile([0.25, 0.75])
long_sleep = df[df["Sleep_Hours"]>= p75]["Exam_Score"].mean()
short_sleep = df[df["Sleep_Hours"] <= p25]["Exam_Score"].mean()

print("Средний балл тех, кто спит больше: ", long_sleep)
print("Средний балл студентов с недосыпом: ", short_sleep)
print("Кто спит больше, сдает экзамен лучше: ", long_sleep > short_sleep)
````

7. Правда ли, что слуденты, чьи родители имеют высшее образование(College || Postgraduate), сдают экзамены лучше, чем остальные?

````python
higher_education = df["Parental_Education_Level"].isin(["College", "Postgraduate"])
has_higher_education = df[higher_education]["Exam_Score"].mean()
others = df[~higher_education]["Exam_Score"].mean()

print("Средний балл у кого родители имеют высшее образование: ", has_higher_education)
print("Средний балл остальных: ", others)
print("Студенты, чьи родители имеют высшее образование, сдают экзамены лучше: ", has_higher_education > others)
````

Однако разница небольшая, и её стоит трактовать с осторожностью: образование родителей — лишь один из многих факторов, а наблюдаемое преимущество может быть опосредовано другими переменными. 

8. Как влияет участие родителей на результаты экзаменов

````python
par_involvement = df.groupby("Parental_Involvement")["Exam_Score"].agg(["mean", "count"]).sort_values("mean")
low_inv = par_involvement.loc["Low", "mean"]
medium_inv = par_involvement.loc["Medium", "mean"]
high_inv = par_involvement.loc["High", "mean"]


print(par_involvement)
print("Участие родителей положительно влияет на успеваемость - средний балл растет с ростом вовлеченности родителей: ", low_inv < medium_inv < high_inv)
````

9. Правда ли, что в семьях с доходом больше среднего дети сдают экзамены лучше?

1. Правда ли, что количество часов затрачиваемых на сон распределено примерно нормально?

````python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))

sns.histplot(df["Sleep_Hours"], bins=20, kde=True, stat="density", color="skyblue")
plt.axvline(df["Sleep_Hours"].mean(), color="red", linestyle="--", label=f"Среднее: {df["Sleep_Hours"].mean():.1f} ч")
plt.legend()
plt.title("Распределение количества часов сна", fontsize=14)
plt.xlabel("Часы сна")
plt.ylabel("Плотность вероятности")
plt.show()

skew = df["Sleep_Hours"].skew()
kurt = df["Sleep_Hours"].kurt()

print(f"Асимметрия (skewness): {skew:.4f}")
print(f"Эксцесс (kurtosis): {kurt:.4f}")

if abs(skew) < 1 and abs(kurt) < 1:
    print("\nВывод: Распределение близко к нормальному (асимметрия и эксцесс близки к 0).")
else:
    print("\nВывод: Распределение имеет отклонения от нормального.")
````

3. Как зависит результат конечного экзамена от результата первого экзамена

````python
plt.figure(figsize=(8, 6))
sns.regplot(data=df, x="Previous_Scores", y="Exam_Score", scatter_kws={"alpha": 0.3, "color": "teal"}, line_kws={"color": "red", "linewidth": 2})

corr = df["Previous_Scores"].corr(df["Exam_Score"])

plt.title("Зависимость итогового балла от первого экзамена", fontsize=14)
plt.xlabel("Балл за первый экзамен (Previous_Scores)")
plt.ylabel("Балл за итоговый экзамен (Exam_Score)")
plt.grid(True, alpha=0.3)
plt.show()

print(f"Коэффициент корреляции: {corr:.3f}")
````

Коэффициент корреляции составляет 0.175, что свидетельствует о крайне слабой линейной связи. На графике рассеяния видно, что облако точек распределено широко, а линия регрессии практически горизонтальна. Это означает, что результат первого экзамена сам по себе не является надежным предиктором итогового балла. Студенты с низкими стартовыми баллами могут отлично сдать финальный экзамен, и наоборот.
